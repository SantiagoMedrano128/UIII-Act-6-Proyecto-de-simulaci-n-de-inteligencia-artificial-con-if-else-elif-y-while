## Crédito automotriz
####Este programa simula un sistema de toma de decisiones de Inteligencia Artificial para la aprobación de un crédito automotriz

```python
# --- SIMULADOR DE IA PARA CRÉDITO AUTOMOTRIZ (AutoIA) ---

print("BIENVENIDO AL SISTEMA DE EVALUACIÓN DE CRÉDITO AutoIA v1.0")
print("---------------------------------------------------------")

# El ciclo while True permite que el programa sea interactivo y se repita [3]
continuar = "s"
while continuar == "s" or continuar == "S":
    # Entrada de datos del usuario [2]
    nombre = input("\nIngrese su nombre completo: ")
    edad = int(input("Ingrese su edad: "))
    ingresos = float(input("Ingrese su sueldo mensual neto: "))
    precio_auto = float(input("Ingrese el precio del auto que desea: "))
    enganche = float(input("¿Cuánto dinero tiene para el enganche?: "))

    print("\nIA-ANALIZANDO DATOS...")

    # Lógica de decisión mediante condicionales e if anidados [1, 4]

    # 1. Verificación de mayoría de edad mediante operador relacional [5]
    if edad < 18:
        print(f"ESTADO: RECHAZADO. {nombre}, debes ser mayor de edad para solicitar un crédito.")

    else:
        # 2. Verificación del enganche mínimo (20% del valor del auto) utilizando operadores aritméticos [2, 6]
        pago_minimo = precio_auto * 0.20

        if enganche < pago_minimo:
            print(f"ESTADO: RECHAZADO. El enganche es insuficiente. El mínimo requerido es {pago_minimo}.")

        else:
            # 3. Cálculo de la capacidad de pago [7, 8]
            # Simulamos un crédito a 48 meses con un interés simple del 10%
            monto_a_financiar = precio_auto - enganche
            interes = monto_a_financiar * 0.10
            pago_mensual = (monto_a_financiar + interes) / 48

            # La IA aprueba si el pago mensual no excede el 30% de los ingresos [9]
            limite_pago = ingresos * 0.30

            if pago_mensual > limite_pago:
                print(f"ESTADO: RECHAZADO. El pago mensual de {pago_mensual:.2f} excede tu capacidad de pago del 30% ({limite_pago:.2f}).")
            elif ingresos < 5000:
                print("ESTADO: RECHAZADO. Tus ingresos son inferiores al mínimo de seguridad para crédito automotriz.")
            else:
                # Caso de aprobación exitosa [10]
                print(f"ESTADO: ¡APROBADO! Felicidades {nombre}.")
                print(f"Detalles del Crédito:")
                print(f"- Monto financiado: {monto_a_financiar:.2f}")
                print(f"- Pago mensual estimado (48 meses): {pago_mensual:.2f}")

    # Pregunta para mantener el ciclo interactivo [11, 12]
    continuar = input("\n¿Desea realizar otra simulación? (s/n): ")

print("\nGracias por utilizar AutoIA. El sistema se ha cerrado.")
```
